# Challenge
Buat skrip Python baru yang mensimulasikan Gerak Parabola (posisi $x$ dan $y$ secara bersamaan) untuk sebuah bola yang ditendang dengan kecepatan awal $v_0 = 30 \text{ m/s}$ pada sudut elevasi $45^\circ$. Plot lintasannya di Matplotlib (sumbu $X$ sebagai jarak horizontal, sumbu $Y$ sebagai ketinggian).

## 1. Parameter Awal

```python
v0 = 30.0 # m/s dengan pada sudut elevasi 45 deg
```

## 2. Menentukan nilai masing-masing sumbu

```python
# Mencari Gerak Parabola Pada  Sumbu Horizontal(X)
# Algoritma Mencari Komponen Sumbu X(Horizontal)
# Mengubah Sudut Elevasi Ke Radian
theta_rad = np.radians(45)
vx = v0 * np.cos(theta_rad)

#  Cari Nilai Variabel Waktu(t) lewat jalur sumbu Vertikal(Y)
g = 9.8
vy = v0 * np.sin(theta_rad)
t_total = 2*vy / g # Waktu total penerbangan (untuk tanah datar)
t = np.linspace(0, t_total, 500)

# Cari Sumbu X
x = vx * t
print(t)
print(vx)
x
```

```python
# Mencari Gerak Parabola Sumbu Vertikal(Y)
# Algoritma Mencari Komponen Sumbu Vertikal(Y)
y = vy * t - 1/2 * g * t**2
y
```

# Visualization

```python
plt.figure(figsize=(8, 5))
plt.plot(x, y, label="Lintasan Parabola", color="blue")
plt.title("Simulasi Gerak Parabola (v0 = 30 m/s, θ = 45°)")
plt.xlabel("Jarak Horizontal x (m)")
plt.ylabel("Ketinggian y (m)")
plt.grid(True)
plt.axhline(0, color="black", linewidth=0.8)
plt.legend()
plt.show()
```
